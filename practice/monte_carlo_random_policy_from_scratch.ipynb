```python
import numpy as np
```

```python
tile_rewards = np.array([[0.02, 0.01, -0.02, 0.03], [0, 0.01, 0.02, 0.01], [0.02, -0.01, 0.01, 0], [0.02, 0.01, 0.02, 0.01]])
```

```python
print(tile_rewards)
```

```python
startX, startY = 1,0
EndX, EndY = 2,3

Actions = ['up','down','left','right']
GameOver = False
```

```python
board = np.zeros([4,4])
print(board)
```

```python
def move(currentPos):
    prevPosition = currentPos
    Action = np.random.choice(Actions)
    if Action == 'up':
        currentPos = (currentPos[0]-1,currentPos[1])
    elif Action == 'down':
        currentPos = (currentPos[0]+1,currentPos[1])
    elif Action == 'left':
        currentPos = (currentPos[0],currentPos[1]-1)
    elif Action == 'right':
        currentPos = (currentPos[0],currentPos[1]+1)
    if (currentPos[0] >= 0) and (currentPos[0] <= 3):
        if (currentPos[1] >= 0) and (currentPos[1] <= 3):
                return currentPos, Action
    return move(prevPosition)
```

```python
print(move((0,0))[1])
```

```python
def checkWin(currentPos):
    if currentPos == (EndX,EndY):
        GameOver = True
        return GameOver
```

```python
def reward(currrentPos):
    if currrentPos == (EndX, EndY):
        return 1
    else:
        return tile_rewards[currrentPos[0],currrentPos[1]]
```

```python
def play(rounds=1):
    i = 0
    currentPosition = (startX, startY)
    singlePolicy = []
    OverallPolicy = []
    while i < rounds:
        if checkWin(currentPosition):
            reward(currentPosition)
            # singlePolicy.append('Win')
            OverallPolicy.append(singlePolicy)
            currentPosition = (startX, startY) # reset
            singlePolicy = [] # reset
            i += 1
        else:
            movement = move(currentPosition)
            currentPosition = movement[0]
            ActionToReachCurrentPosition = movement[1]
            reward(currentPosition)
            # print((currentPosition, ActionToReachCurrentPosition, reward(currentPosition)))
            singlePolicy.append((currentPosition, ActionToReachCurrentPosition, reward(currentPosition)))
    return OverallPolicy
```

```python
play(50)
```
